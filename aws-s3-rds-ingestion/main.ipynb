```python
import psycopg2
import boto3
import io
import os
from dotenv import load_dotenv

load_dotenv()

s3 = boto3.resource(
    service_name='s3',
    region_name=os.getenv("AWS_REGION"),
    aws_access_key_id=os.getenv("AWS_ACCESS_KEY_ID"),
    aws_secret_access_key=os.getenv("AWS_SECRET_ACCESS_KEY")
)

bucket_name = os.getenv("S3_BUCKET_NAME")
prefix = 'image/'

con = psycopg2.connect(
    host=os.getenv("RDS_HOST"),
    port=os.getenv("RDS_PORT"),
    database="inventory",
    user=os.getenv("RDS_USER"),
    password=os.getenv("RDS_PASSWORD")
)
con.autocommit = True
cursor = con.cursor()
id = 0

for obj in s3.Bucket(bucket_name).objects.filter(Prefix=prefix):
    if obj.key.endswith('.jpg') or obj.key.endswith('.JPG') or obj.key.endswith('.jpeg') or obj.key.endswith('.JPEG') or obj.key.endswith('.png') or obj.key.endswith('.PNG'):
        file_name = obj.key.split('/')[1]
        #print(file_name)
        id += 1 

        cursor.execute(
    "INSERT INTO files (idfiles, filename) VALUES (%s, %s)",
    (id, file_name)
)
        print(f"Inserted {file_name} into RDS table.")
```

```python
cursor.execute("SELECT * FROM files;")
var = cursor.fetchall()
for row in var:
    print(row)
cursor.close()
```
