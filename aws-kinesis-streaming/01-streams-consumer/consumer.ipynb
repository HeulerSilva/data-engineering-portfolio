```python
import boto3
import os
from dotenv import load_dotenv

load_dotenv()

client = boto3.client(
    service_name='kinesis',
    region_name=os.getenv("AWS_REGION"),
    aws_access_key_id=os.getenv("AWS_ACCESS_KEY_ID"),
    aws_secret_access_key=os.getenv("AWS_SECRET_ACCESS_KEY")
)

shard = client.get_shard_iterator(
    StreamName=os.getenv("KINESIS_STREAM_NAME"),
    ShardId=os.getenv("KINESIS_SHARD_ID"),
    ShardIteratorType="LATEST"
)["ShardIterator"]

max_records = 3
count = 0

while shard is not None and count < max_records:
    response = client.get_records(ShardIterator=shard, Limit=10)
    records = response["Records"]
    for record in records:
        print(record["SequenceNumber"])
        print(record["ApproximateArrivalTimestamp"])
        print(record["PartitionKey"])
        print(record["Data"].decode("utf-8"))
        count += 1
        if count >= max_records:
            break
    shard = response.get("NextShardIterator")

print(f"Encerrado após processar {count} registro(s).")
```
