```python
import boto3
import json
import os
from dotenv import load_dotenv

load_dotenv()

client = boto3.client(
    service_name='kinesis',
    region_name=os.getenv("AWS_REGION"),
    aws_access_key_id=os.getenv("AWS_ACCESS_KEY_ID"),
    aws_secret_access_key=os.getenv("AWS_SECRET_ACCESS_KEY")
)

#register = {'idseller': '995', 'name': 'Jane Damo', 'email': 'janedamo@example.com'}
#register = {'idseller': '994', 'name': 'Craig Johnson', 'email': 'craigjohnson@example.com'}
register = {'idseller': '993', 'name': 'William Smith', 'email': 'williamsmith@example.com'}

response = client.put_record(
    StreamName=os.getenv("KINESIS_STREAM_NAME"),
    Data=json.dumps(register),
    PartitionKey=os.getenv("KINESIS_PARTITION_KEY")
)

print(f"Record sent to Kinesis stream. Sequence number: {response['SequenceNumber']}")
```
